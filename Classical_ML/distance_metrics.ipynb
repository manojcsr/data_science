```python
from scipy.spatial import distance

# two sample vectors
a = [1, 2, 3]
b = [4, 6, 8]

# Euclidean distance (L2)
eu = distance.euclidean(a, b)
print("Euclidean:", eu)

# Manhattan distance (L1, aka cityblock)
man = distance.cityblock(a, b)
print("Manhattan:", man)

# Minkowski distance (general form, p=3 here)
mink = distance.minkowski(a, b, p=3)
print("Minkowski (p=3):", mink)

# Hamming distance (fraction of differing elements)
ham = distance.hamming([1, 0, 1, 1], [1, 1, 0, 1]) 
print("Hamming:", ham, "( 2 differences out of 4 )")

x1 = "Hello"
x2 = "World"
print(distance.hamming(list(x1), list(x2)), "(4 mismatches out of 5 → 4/5 = 0.8)") 
```
