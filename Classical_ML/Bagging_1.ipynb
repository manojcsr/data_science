# Bagging - Bootstrapped Aggregating

```python
from sklearn.ensemble import BaggingClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

# Load data
X, y = load_wine(return_X_y=True)
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=42)

# Base model = Decision Tree
base_tree = DecisionTreeClassifier(random_state=42)

# Bagging model
bag_model = BaggingClassifier(
    estimator=base_tree,
    n_estimators=100,
    bootstrap=True,
    n_jobs=-1,
    random_state=42
)
bag_model.fit(X_train, y_train)

# Evaluate
y_pred = bag_model.predict(X_test)
print("Bagging Accuracy:", accuracy_score(y_test, y_pred))
```

```python
from sklearn.tree import DecisionTreeClassifier

# Load data
X, y = load_wine(return_X_y=True)
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=42)

dt = DecisionTreeClassifier(random_state=42)
dt.fit(X_train, y_train)
y_pred = dt.predict(X_test)

print("Decision tree Accuracy:", accuracy_score(y_test, y_pred))
```

```python
# Bagging with KNN example
import numpy as np
import pandas as pd
from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report
from sklearn.ensemble import BaggingClassifier
from sklearn.neighbors import KNeighborsClassifier
import matplotlib.pyplot as plt

# 1️⃣ Load dataset
X, y = load_wine(return_X_y=True)

# 2️⃣ Train-test split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42, stratify=y
)

# 3️⃣ Base model: KNN
knn = KNeighborsClassifier(n_neighbors=5)

# 4️⃣ Bagging with KNN as base estimator
bag_knn = BaggingClassifier(
    estimator=knn,            # base model
    n_estimators=5,          # number of KNN models
    bootstrap=True,           # sample with replacement
    n_jobs=-1,                # parallel processing
    random_state=42
)

# 5️⃣ Train both models
knn.fit(X_train, y_train)
bag_knn.fit(X_train, y_train)

# 6️⃣ Evaluate
y_pred_knn = knn.predict(X_test)
y_pred_bag = bag_knn.predict(X_test)

acc_knn = accuracy_score(y_test, y_pred_knn)
acc_bag = accuracy_score(y_test, y_pred_bag)

print(f"Single KNN Accuracy: {acc_knn:.3f}")
print(f"Bagging KNN Accuracy: {acc_bag:.3f}")

print("\n--- Single KNN Report ---")
print(classification_report(y_test, y_pred_knn, digits=3))

print("\n--- Bagging KNN Report ---")
print(classification_report(y_test, y_pred_bag, digits=3))
```

```python
from sklearn.datasets import make_moons
from sklearn.inspection import DecisionBoundaryDisplay

# Create a 2D non-linear dataset
X, y = make_moons(n_samples=400, noise=0.3, random_state=42)
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=42)

# Base and bagged models
knn = KNeighborsClassifier(n_neighbors=5)
bag_knn = BaggingClassifier(estimator=knn, n_estimators=25, random_state=42)

knn.fit(X_train, y_train)
bag_knn.fit(X_train, y_train)

# Plot decision boundaries
fig, ax = plt.subplots(1, 2, figsize=(10, 4))
DecisionBoundaryDisplay.from_estimator(knn, X, response_method="predict", ax=ax[0])
DecisionBoundaryDisplay.from_estimator(bag_knn, X, response_method="predict", ax=ax[1])
ax[0].set_title("Single KNN")
ax[1].set_title("Bagging (KNN Ensemble)")
plt.tight_layout()
plt.show()
```

KNN gives a jagged, overfitted boundary.

Bagging KNN gives a smoother, more general decision surface.

# Code Demo - Random Forest

```python
# Decision Tree vs Random Forest on Wine Quality (multi-class)
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from collections import Counter

from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.metrics import accuracy_score, classification_report, ConfusionMatrixDisplay
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier

# 1) Load dataset (UCI: red wine quality)
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/wine-quality/winequality-red.csv"
df = pd.read_csv(url, sep=';')

X = df.drop(columns=['quality'])
y = df['quality']  # multiclass labels: usually {3,4,5,6,7,8}

print("Dataset shape:", df.shape)
print("Class distribution:", Counter(y))

# 2) Train/Test split (stratified)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42, stratify=y
)

# 3) Baseline Decision Tree
dt = DecisionTreeClassifier(random_state=42)
dt.fit(X_train, y_train)
y_pred_dt = dt.predict(X_test)

print("\n=== Decision Tree (baseline) ===")
print("Accuracy:", accuracy_score(y_test, y_pred_dt))
print(classification_report(y_test, y_pred_dt, digits=3))

# 4) Baseline Random Forest
rf = RandomForestClassifier(n_estimators=300, random_state=42, n_jobs=-1)
rf.fit(X_train, y_train)
y_pred_rf = rf.predict(X_test)

print("\n=== Random Forest (baseline) ===")
print("Accuracy:", accuracy_score(y_test, y_pred_rf))
print(classification_report(y_test, y_pred_rf, digits=3))

# 5) Confusion matrices side-by-side
fig, ax = plt.subplots(1, 2, figsize=(12, 5))
ConfusionMatrixDisplay.from_predictions(y_test, y_pred_dt, ax=ax[0])
ax[0].set_title("Decision Tree")
ConfusionMatrixDisplay.from_predictions(y_test, y_pred_rf, ax=ax[1])
ax[1].set_title("Random Forest")
plt.tight_layout()
plt.show()

# 6) (Optional) Quick tuning for better generalization
#    You can comment this out if you want just the baselines.
dt_grid = {"max_depth": [3, 5, 7, 9, None],
           "min_samples_leaf": [1, 3, 5, 10],
           "criterion": ["gini", "entropy", "log_loss"]}

rf_grid = {"n_estimators": [200, 400],
           "max_depth": [None, 10, 20],
           "min_samples_leaf": [1, 2, 4],
           "max_features": ["sqrt", "log2", None]}

dt_cv = GridSearchCV(DecisionTreeClassifier(random_state=42),
                     dt_grid, scoring="accuracy", cv=5, n_jobs=-1)
dt_cv.fit(X_train, y_train)
best_dt = dt_cv.best_estimator_
y_pred_dt_best = best_dt.predict(X_test)

rf_cv = GridSearchCV(RandomForestClassifier(random_state=42, n_jobs=-1),
                     rf_grid, scoring="accuracy", cv=5, n_jobs=-1)
rf_cv.fit(X_train, y_train)
best_rf = rf_cv.best_estimator_
y_pred_rf_best = best_rf.predict(X_test)

print("\n=== Tuned Decision Tree ===")
print("Best params:", dt_cv.best_params_)
print("Accuracy:", accuracy_score(y_test, y_pred_dt_best))
print(classification_report(y_test, y_pred_dt_best, digits=3))

print("\n=== Tuned Random Forest ===")
print("Best params:", rf_cv.best_params_)
print("Accuracy:", accuracy_score(y_test, y_pred_rf_best))
print(classification_report(y_test, y_pred_rf_best, digits=3))

# 7) Feature importances (from tuned RF)
fi = pd.Series(best_rf.feature_importances_, index=X.columns).sort_values(ascending=True)
plt.figure(figsize=(8, 5))
fi.plot(kind="barh")
plt.title("Feature Importances — Tuned Random Forest")
plt.xlabel("Importance")
plt.tight_layout()
plt.show()
```

```python
# California Housing Regression using Decision Tree and Random Forest
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import fetch_california_housing
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

# 1️⃣ Load dataset
housing = fetch_california_housing(as_frame=True)
X = housing.data
y = housing.target  # median house value (in $100,000s)

print("Shape:", X.shape)
print("Feature names:", list(X.columns))

# 2️⃣ Train-test split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42
)

# 3️⃣ Decision Tree Regressor
dt = DecisionTreeRegressor(random_state=42)
dt.fit(X_train, y_train)
y_pred_dt = dt.predict(X_test)

# 4️⃣ Random Forest Regressor
rf = RandomForestRegressor(
    n_estimators=300, random_state=42, n_jobs=-1
)
rf.fit(X_train, y_train)
y_pred_rf = rf.predict(X_test)

# 5️⃣ Evaluation function
def eval_model(name, y_true, y_pred):
    rmse = np.sqrt(mean_squared_error(y_true, y_pred))
    mae = mean_absolute_error(y_true, y_pred)
    r2 = r2_score(y_true, y_pred)
    print(f"\n=== {name} ===")
    print(f"RMSE: {rmse:.3f}")
    print(f"MAE : {mae:.3f}")
    print(f"R²  : {r2:.3f}")
    return rmse, mae, r2

rmse_dt, mae_dt, r2_dt = eval_model("Decision Tree Regressor", y_test, y_pred_dt)
rmse_rf, mae_rf, r2_rf = eval_model("Random Forest Regressor", y_test, y_pred_rf)

# 6️⃣ Feature importances (from RF)
fi = pd.Series(rf.feature_importances_, index=X.columns).sort_values(ascending=True)
plt.figure(figsize=(8, 5))
fi.plot(kind='barh')
plt.title("Feature Importances — Random Forest Regressor")
plt.xlabel("Importance")
plt.tight_layout()
plt.show()

# 7️⃣ Optional — Hyperparameter tuning for better generalization
param_grid_dt = {
    "max_depth": [5, 8, 12, None],
    "min_samples_leaf": [1, 3, 5, 10],
    "criterion": ["squared_error", "friedman_mse"]
}

grid_dt = GridSearchCV(
    DecisionTreeRegressor(random_state=42),
    param_grid_dt, cv=5, scoring="r2", n_jobs=-1
)
grid_dt.fit(X_train, y_train)

print("\nBest Decision Tree Params:", grid_dt.best_params_)
best_dt = grid_dt.best_estimator_
y_pred_best_dt = best_dt.predict(X_test)
eval_model("Tuned Decision Tree Regressor", y_test, y_pred_best_dt)
```

# FAQs - for Decision Trees and Random Forest

<b> Do you need scaling? </b>>

No, you don’t need scaling.

Decision Trees and Random Forests don’t care about feature scale.

Whether your data is in meters, kilometers, or millions — it doesn’t matter.

These models split data based on thresholds (like “is feature < 10?”), not on distances or gradients.

👉 Example: If one column is “income in ₹” and another is “age in years,” the model handles both just fine — no normalization or standardization needed.

<b> Do you need to remove outliers before training? </b>

Not necessarily — they are quite robust, but…

Decision Trees and Random Forests can handle outliers better than models like Linear Regression or SVM.

That’s because a single extreme value only affects one split in the tree, not the whole model.

However, if your data has many extreme or incorrect outliers, it can still confuse the model (especially small trees).

👉 So: You don’t have to remove outliers, but it’s good practice to check and clean obvious mistakes.

<b> Do you need to impute (fill in) missing values? </b>

Yes, it’s recommended.

Most Decision Tree and Random Forest implementations (like in scikit-learn) don’t accept missing values directly.

You can fill them with mean, median, mode, or other techniques before training.

👉 So: You should impute missing data before training, or drop rows/columns if appropriate.

<b> Can you apply regularization like Lasso or Ridge? </b>

No, not directly.

Lasso and Ridge are designed for linear models (like Linear or Logistic Regression).

Trees don’t use weights like those models — they split data into groups instead.

But you can control overfitting in trees using hyperparameters, which act as a kind of “tree regularization”:

max_depth → limits how deep the tree can grow

min_samples_leaf → minimum number of samples per leaf

max_features → limits number of features used per split (used in Random Forest)

👉 These parameters regularize the model, even though it’s not Lasso or Ridge.

# Using OOB score to evaluate Random Forest

```python
# Random Forest Regression on California Housing with OOB score
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import fetch_california_housing
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

# 1) Load dataset
housing = fetch_california_housing(as_frame=True)
X = housing.data
y = housing.target  # median house value (in $100,000s)

print("Shape:", X.shape)
print("Feature names:", list(X.columns))

# 2) Train-test split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42
)

# 3) Random Forest with OOB enabled
#    - oob_score=True requires bootstrap=True (default)
#    - oob_score_ is an R^2 estimate computed on each tree's "out-of-bag" samples
rf = RandomForestRegressor(
    n_estimators=400,
    oob_score=True,        # <-- OOB evaluation
    random_state=42,
    n_jobs=-1
)

rf.fit(X_train, y_train)

# 4) OOB performance (training-time, unbiased-ish estimate)
print(f"OOB R^2 (training estimate): {rf.oob_score_:.3f}")

# 5) Test-set predictions & metrics
y_pred = rf.predict(X_test)

rmse = np.sqrt(mean_squared_error(y_test, y_pred))
mae  = mean_absolute_error(y_test, y_pred)
r2   = r2_score(y_test, y_pred)

print("\n=== Test-set Performance ===")
print(f"RMSE: {rmse:.3f}")
print(f"MAE : {mae:.3f}")
print(f"R^2 : {r2:.3f}")
```
