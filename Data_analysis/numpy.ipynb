# boolean masking

```python
arr = np.array([10, 20, 30, 40, 50])
```

```python
mask = arr > 30
```

```python
print(mask)
```

```python
print(arr[mask])
```

```python
print(arr[(arr > 20) & (arr < 50)])
```

```python
mat = np.array([[5, 10, 15],
                [20, 25, 30],
                [35, 40, 45]])

mask = mat > 20
print(mask)
print(mat[mask])
```

```python
arr = np.array([10, 20, 30, 40, 50])
arr[arr > 30] = 999
print(arr)
```

```python
result = np.where(arr > 30, "High", "Low")
print(result)
```

```python
mat = np.array([[1, 8, 3], [4, 5, 6], [0, 2, 3]])
from pprint import pprint
pprint(mat)

# row mask
row_mask = np.any(mat > 4, axis=1)
pprint(mat[row_mask])

# col mask
col_mask = np.any(mat > 4, axis=0)
filtered = mat[:, col_mask]
pprint(filtered)
```

# Reshaping arrays

```python
# Reshaping means changing the structure of a NumPy array (e.g., from 1D to 2D, 2D to 3D, etc.) 
# while keeping all the original elements.
```

```python
# np.reshape(array, new_shape)
```

```python
# Image processing - Reshape flat pixel data into 2D or 3D arrays
# Matrix operations - Many require specific dimensionality
```

```python
arr = np.arange(6)
print(arr)
reshaped = arr.reshape((2, 3))
print(reshaped)
```

```python
arr2 = np.array([[1, 2, 3], [4, 5, 6]])
flattened = arr2.reshape(-1) # flatten into 1D
print(flattened)
```

```python
arr3 = np.arange(12)
print(arr3)
reshaped = arr3.reshape(3, -1)  # -1 allows NumPy to figure out the second dim (12 ÷ 3 = 4)
print(reshaped)
```

```python
np.arange(10).reshape((3, 4))
```

```python
# memory efficient flattening

a = np.array([[1, 2, 3], [4, 5, 6]])
r = np.ravel(a) # other method is np.flatten
print(r)
```

```python
b = a.T
print(b)
print(np.ravel(b))
```

```python
x = np.arange(12).reshape(3, 4)
print(x)
print(np.ravel(x))
print(np.ravel(x.T))
```

# Stacking arrays

```python
# Stacking means joining arrays together along a new or existing axis.
# It’s like stacking Lego blocks — either vertically (on top of each other), horizontally (side by side)
```

```python
a = np.array([[1, 2], [3, 4]])
b = np.array([[5, 6], [7, 8]])
```

```python
print(a)
print(b)
v = np.vstack((a, b))
print(v)
```

```python
h = np.hstack((a, b))
print(h)
```

```python
c = np.concatenate((a, b), axis=0) # same as vstack
print(c)
```

# Splitting arrays

```python
arr = np.arange(16).reshape(4, 4)
print(arr)
```

```python
upper, lower = np.split(arr, 2, axis=0)  # Split rows
print(upper)
print(lower)
```

```python
left, right = np.split(arr, 2, axis=1)  # Split columns
```

```python
print(left)
```

```python
print(right)
```

```python
print(arr)
left, right = np.hsplit(arr, 2) # same as split columns above
print(left)
print(right)
```

```python
top, bottom = np.vsplit(arr, 2) # same as Split rows above
print(top)
print(bottom)
```

# File I/O

# Copy vs view 

```python
arr = np.array([1, 2, 3])
arr_copy = arr.copy()

arr_copy[0] = 99

print("Original:", arr)      
print("Copy:    ", arr_copy) 
```

```python
arr = np.array([1, 2, 3])
arr_view = arr.view()

arr_view[0] = 99

print("Original:", arr)     
print("View:    ", arr_view) 
```
