```python
import matplotlib.pyplot as plt

labels = ['Math', 'Science', 'English']
male = [20, 15, 5]
female = [10, 20, 30]

x = range(len(labels))

plt.bar(x, male, label='Male')
plt.bar(x, female, bottom=male, label='Female')

plt.xticks(x, labels)
plt.ylabel('Number of Students')
plt.title('Favorite Subjects by Gender')
plt.legend()
plt.show()
```

```python
# Scatter
```

```python
import matplotlib.pyplot as plt

# Sample data
hours = [1, 2, 3, 4, 5, 6, 7, 8, 9]
scores = [52, 55, 60, 65, 70, 75, 80, 85, 90]

# Create scatter plot
plt.scatter(hours, scores)
plt.title("Scatter Plot of Hours Studied vs Test Score")
plt.xlabel("Hours Studied")
plt.ylabel("Test Score")
plt.grid(True)
plt.show()
```

```python
# Timeseries
```

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun", 
          "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]
sales = [5000, 5200, 6000, 5800, 6200, 6300, 6400, 6500, 6700, 6800, 7000, 7500]

plt.plot(months, sales, marker='o')
plt.title("Monthly Bookstore Sales in 2024")
plt.xlabel("Month")
plt.ylabel("Sales (in ₹)")
plt.grid(True)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Create subplots
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Sample data
x = np.arange(1, 6)
y = np.array([2, 3, 5, 7, 11])

# Create 2x2 subplots
fig, axs = plt.subplots(2, 2, figsize=(10, 8))

# 1. Line Plot
axs[0, 0].plot(x, y, marker='o')
axs[0, 0].set_title("Line Plot")
axs[0, 0].set_xlabel("X")
axs[0, 0].set_ylabel("Y")

# 2. Bar Chart
axs[0, 1].bar(x, y, color='skyblue')
axs[0, 1].set_title("Bar Chart")

# 3. Scatter Plot
axs[1, 0].scatter(x, y, color='green')
axs[1, 0].set_title("Scatter Plot")

# 4. Histogram
random_data = np.random.randn(100)
axs[1, 1].hist(random_data, bins=10, color='orange')
axs[1, 1].set_title("Histogram")

# Layout adjustment
plt.tight_layout()
plt.show()
```
