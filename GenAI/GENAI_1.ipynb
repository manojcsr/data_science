Encoder block contains a self-attention layer and a feedforward neural network

Self-attention attends to all parts of the input sequence at once so that it can "look" both forward and backward in a single sequence. And it does in parallel.

Decoder has additional attention layer. Self-attention layer in the decoder masks future positions so it only attends to earlier positions so it only attends to earlier positions to prevent leaking.  The decoder is NOT allowed to look at future words. Because during generation, the decoder should predict the next word without cheating.

Without masking, the decoder would “peek” at future words during training and learn unrealistically.

# Representation Models : Encoder-only models

The original Transformer model is an encoder-decoder architecture that serves translation tasks well but cannnot easily be used for other tasks, like text classification.

Representation models focus only on representation (embeddings) but DO NOT generate text.

In 2018, BERT (Bidirectional Encoder Representations fro Transfomers) was introduced.

Bidirectional encoder - it understands :

the word itself, the words before it, and the words after it

to build a deep contextual representation of every token.

<img src="./images/genai_13.png" width=1000 />

Input contains an additional token (CLS) or classification token which is used as representation for the entire input.

Training the encoder is done by masked language modelling. This prediction task is difficult but allows BERT to create more accurate (intermediate) representations of the input.

```python
from transformers import AutoTokenizer, AutoModelForCausalLM
import torch

device = "mps" if torch.backends.mps.is_available() else "cpu"

tokenizer = AutoTokenizer.from_pretrained("microsoft/Phi-3-mini-4k-instruct")
model = AutoModelForCausalLM.from_pretrained(
    "microsoft/Phi-3-mini-4k-instruct",
    torch_dtype=torch.float16,
).to(device)
```

### Model properties

Phi-3 Mini 4K Instruct is a 3.8B-parameter, instruction-tuned small language model (SLM) from Microsoft’s Phi-3 family.

It is optimized for:

<pre>
	•	High quality on reasoning relative to size
	•	Very low latency
	•	Running efficiently on edge devices
	•	Small VRAM budgets (6–8 GB GPUs)
	•	Short-context tasks (up to 4K tokens)
</pre>

```python
# A chat conversation is represented as a list of message dictionaries.
# Each message has a "role" (user / assistant / system) and "content" (text).
# This structure matches the format the Phi-3 model expects for chat tasks.
messages = [
    {"role": "user", "content": "Tell me a joke on Artificial Intelligence !"}
]


# ----------------------------------------------------------------------------------
# Convert the conversation into model-understandable token IDs
# using the model’s chat template.
# 
# WHY WE USE apply_chat_template:
# - Chat models (Phi, LLaMA, Mistral-Instruct) require a specific *prompt format*
#   with special tokens like <s>, </s>, <|assistant|>, <|user|>, etc.
# - apply_chat_template automatically adds all these special tokens.
#
# add_generation_prompt=True:
# - This tells the tokenizer: 
#     “We have finished feeding the user messages:
#      now insert the special ‘assistant turn’ token,
#      and prepare the model to start generating the assistant’s reply.”
#
# Without add_generation_prompt=True:
# - The model may continue speaking in the *user role* or output garbage.
#
# return_tensors="pt":
# - Ask tokenizer to return PyTorch tensors instead of Python lists.
#
# .to(device):
# - Moves the tensor to the correct device (CPU, MPS, GPU).
# ----------------------------------------------------------------------------------

input_ids = tokenizer.apply_chat_template(
    messages,
    add_generation_prompt=True,
    return_tensors="pt"
).to(device)

print(input_ids)

for id in input_ids[0]:
    print(id, tokenizer.decode(id))
# ----------------------------------------------------------------------------------
# GENERATION PHASE
#
# model.generate():
# - Performs auto-regressive next-token generation.
# - Takes the prompt (input_ids) and starts generating new tokens until
#   a stopping condition is reached.
#
# torch.no_grad():
# - Disables gradient tracking since we're doing inference, not training.
# - Saves memory and increases inference speed.
#
# max_new_tokens=200:
# - Limits how much text the model can generate.
# - Prevents runaway infinite loops.
#
# temperature=0.7:
# - Controls randomness.
#   Lower = deterministic, Higher = more creative.
#
# top_p=0.9:
# - Nucleus sampling. Restricts token choices to the smallest set whose
#   cumulative probability ≥ 0.9. Produces human-like text.
#
# do_sample=True:
# - Enables sampling instead of greedy decoding.
# - Required when using temperature or top_p.
#
# outputs:
# - A tensor containing BOTH:
#     [the original prompt tokens] + [newly generated tokens]
# ----------------------------------------------------------------------------------

with torch.no_grad():
    outputs = model.generate(
        input_ids,
        max_new_tokens=200,
        temperature=0.1,
        top_p=0.1,
        do_sample=True,
    )


# ----------------------------------------------------------------------------------
# EXTRACT ONLY THE NEW TOKENS (THE ASSISTANT'S REPLY)
#
# outputs[0]             → The generated sequence (prompt + reply).
# input_ids.shape[-1]    → Length of the original prompt.
#
# outputs[0, input_len:] → Slice off the prompt; keep only new tokens.
#
# This is CRITICAL because:
# - generate() returns the full sequence including the input.
# - If we decode everything, we will print the prompt again mixed with the answer.
# - This slicing ensures we decode only the assistant-generated continuation.
# ----------------------------------------------------------------------------------

generated_ids = outputs[0, input_ids.shape[-1]:]
print(generated_ids)

# ----------------------------------------------------------------------------------
# DECODE TOKENS BACK INTO HUMAN-READABLE TEXT
#
# tokenizer.decode():
# - Converts token IDs → string.
#
# skip_special_tokens=True:
# - Removes tokens like <s>, </s>, <|assistant|>, etc.
# - Produces clean output suitable for display.
# ----------------------------------------------------------------------------------
for id in generated_ids:
    print(id, tokenizer.decode(id))
response = tokenizer.decode(generated_ids, skip_special_tokens=True)

print(response)
```

##### Sampling decoding

A better approach is to add some randomness and sometimes choose the second or third highest probability token. 

Idea here is to basically sample from the probability distribution based on the probability score.

This is the do_sample parameter we have seen above

```python
# We create a chat-style input for an instruction-tuned model.
# "messages" mimics a conversation where the user says:
#     "The capital of India is"
# The model is expected to complete this in an autoregressive manner.
messages = [
    {"role": "user", "content": "The capital of India is"},
]

# Select the device for running the model.
# "mps" = Apple's GPU backend (for M1/M2/M3 chips).
# Otherwise we fall back to CPU.
device = "mps" if torch.backends.mps.is_available() else "cpu"


# -------------------------------------------------------------------
# 1. APPLY CHAT TEMPLATE → TOKENIZE THE INPUT
# -------------------------------------------------------------------
# tokenizer.apply_chat_template() takes the messages and converts them
# into the model’s expected chat formatting (like adding system/user
# markers, prompt tokens, special tokens, etc.).
#
# tokenize=True: convert text → token IDs.
# add_generation_prompt=True: inserts the "assistant starts speaking now"
#     token so the model knows to continue generation after the user query.
#
# return_tensors="pt": return PyTorch tensors.
#
# The output 'enc' is a tensor of shape [1, seq_len] containing token IDs.
# These token IDs become input to the decoder-only transformer.
enc = tokenizer.apply_chat_template(
    messages,
    tokenize=True,
    add_generation_prompt=True,
    return_tensors="pt",
).to(device)

print(enc.shape)

# -------------------------------------------------------------------
# 2. AUTOREGRESSIVE GENERATION LOGIC (INSIDE model.generate)
# -------------------------------------------------------------------
# The generate(...) function performs autoregressive decoding.
# "Autoregressive" means:
#     The model predicts ONE token at a time,
#     and each newly predicted token becomes part of the input for the
#     next prediction.
#
# max_new_tokens=10:
#   - The model will generate up to 10 new tokens beyond the initial prompt.
#
# do_sample=False:
#   - Greedy decoding (always pick highest probability token).
#
# HOW GENERATE WORKS INTERNALLY:
# -------------------------------------------------------------------
# Step A: Feed the input sequence (enc) into the transformer.
#         The model produces hidden states for each input token.
#
# Step B: The final hidden state for the **last token**
#         is passed through the LM HEAD (a linear layer mapping
#         hidden_size → vocab_size).
#
#         LM HEAD OPERATION:
#             logits = lm_head(last_hidden_state)
#
#         This gives a vector of size [vocab_size] that represents the
#         probability distribution over ALL possible next tokens.
#         Example: logits[32064] = score for each of the 32,064 tokens.
#
# Step C: Since do_sample=False, the model picks the token with the
#         highest logit value (argmax).
#
# Step D: Append the newly chosen token to the input sequence.
#         The input length increases by 1 (seq_len → seq_len+1).
#
# Step E: Repeat Steps A–D again:
#         The model recomputes only for the NEW token using KV-cache.
#         Again apply LM head → pick next token → append → repeat.
#
# This loop continues until the model generates 10 new tokens
# or hits a stop token.
generated = model.generate(enc, max_new_tokens=4, do_sample=False)

print(generated, generated.shape)

# -------------------------------------------------------------------
# 3. DECODE THE GENERATED TOKEN IDs BACK TO TEXT
# -------------------------------------------------------------------
# generated has shape [1, original_seq_len + generated_len].
# We decode only generated[0], and skip special tokens (like <s>, </s>).
#
# The output contains:
#     - The user prompt (formatted by chat template)
#     - The assistant’s continuation, autoregressively generated
#
print(tokenizer.decode(generated[0], skip_special_tokens=True))
```

### Parallel token processing

<img src="./images/genai_50.png" width=1000 />

Tokens gets processed parallel initially.

<pre>

    In a Transformer, self-attention processes all input tokens in parallel during a forward pass.

    ["The", "capital", "of", "India", "is"]

    Token embeddings (vectors) for all 5 tokens are fed into the model together.

    Q, K, V matrices are computed in parallel

    Attention weights between every pair of tokens are computed in parallel

    Feed-forward networks process each position in parallel

    This is why Transformers can use GPUs efficiently.

</pre>

Current Transformer models have a limit for how many tokens they can process at once. That limit is called model's context length.

Each of the token streams starts with token embedding and positional embedding. Another vector emerges as the result of the model's processing called contextual embedding

For text generation, only the output result of the last stream is used to preidct the next token.

That output vector is the only input into the LM head as it calculates the probabilities of next token.

### Q,K,V matrix and attention score

“Sarah fed the cat because it…”

Stage 1 — Each word starts with a token embedding

Stage 2 — Self-attention transforms token embeddings → contextual embeddings

=====

Attention is all you need - A way to score how relevant each of the previous input tokens are to the current token being processed. 

Using these scores, we combine the informaiton from previous positions into a single output vector.

##### How attention is calculated ?

(1) Compute Q, K, V for each token

Q = embedding × Wq  
K = embedding × Wk  
V = embedding × Wv

Where Wq, Wk, Wv are learned matrices.


(2) Compute attention scores - the attention score is computed for every pair of tokens in the sentence.

scores = (Q · Kᵀ) / √d_k

<pre>
      T1    T2    T3    T4    T5
T1   s11   s12   s13   s14   s15
T2   s21   s22   s23   s24   s25
T3   s31   s32   s33   s34   s35
T4   s41   s42   s43   s44   s45
T5   s51   s52   s53   s54   s55

s(i,j) = similarity score between Qi and Kj

</pre>

(3) Apply masking (in decoder) so token t sees only tokens ≤ t

scores[j] = -∞ for j > t

<pre>
        T1   T2   T3   T4   T5
T1      ✓    X    X    X    X
T2      ✓    ✓    X    X    X
T3      ✓    ✓    ✓    X    X
T4      ✓    ✓    ✓    ✓    X
T5      ✓    ✓    ✓    ✓    ✓
</pre>

(4) Softmax converts scores → weights

weights = softmax(scores)

(5) Create contextual embedding using weighted sum of V

contextual_embedding = Σ weights[i] × V[i]

This final contextual embedding is passed to:

<pre>
	•	next transformer layer
	•	or LM head if it’s the last layer

</pre>

### Positional embddings

Transformers do not have any inherent notion of order.
<pre>

Unlike RNNs:
	•	They do not process tokens sequentially
	•	Self-attention looks at all tokens in parallel

</pre>

Without positional embeddings, the model sees the sentence “India is beautiful” and “Beautiful is India” as identical bags of tokens.

Positional embeddings inject order information into token embeddings. 

This positional information flows through every transformer block, shaping attention patterns, contextual embeddings, and ultimately the next-token prediction.

Each token embedding is added to a positional embedding:

input_vector[i] = token_embedding[i] + positional_embedding[i]

This combined vector is the actual input to Transformer Block 1.

<pre>

Only positional embeddings allow the model to detect:
	•	subject
	•	verb
	•	object
	•	syntactic structure

</pre>
