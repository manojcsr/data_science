### Task specific model

```python
from transformers import pipeline
```

```python
model_path = 'cardiffnlp/twitter-roberta-base-sentiment-latest'
```

```python
pipe = pipeline(
    model=model_path,
    tokenizer=model_path,
    return_all_scores=True,
    device='mps'
)
```

<img src="./images/genai_57.png" width=1000 />

```python
import numpy as np
```

```python
from tqdm import tqdm
```

```python
from transformers.pipelines.pt_utils import KeyDataset
```

```python
y_pred = []
```

```python
results = tqdm(
    pipe(
        KeyDataset(
            data["test"], "text"
        )
    ),
    total=len(data["test"])
)
for output in results:
    print(output)
```

```python
for output in results:
    negative_score = output[0]["score"]
    positive_score = output[2]["score"]
    assignment = np.argmax([negative_score,positive_score])
    y_pred.append(assignment)
```

```python
len(y_pred)
```

```python
from sklearn.metrics import classification_report
```

```python
def evaluate_performance(y_true, y_pred):
    performance = classification_report(
        y_true,y_pred,
        target_names=["Negative review", "Positive review"]
    )
    print(performance)
```

```python
evaluate_performance(data["test"]["label"], y_pred)
```

### Embedding approach

In the previous example, we used a pretrained task-specific model for sentiment analysis.  

What if we cannot find a model that was pre-trained for a specific task ? We can fine-tune but do we have compute for it ?

That's where general purpose embedding models come in.

<img src="./images/genai_58.png" width=1000 />

We convert our textual input to embedding using the embedding model. Note that this model is kept frozen and is not updated during the training process.

```python
from sentence_transformers import SentenceTransformer
```

```python
model = SentenceTransformer("sentence-transformers/all-mpnet-base-v2")
```

##### Sentence embedding is way better than word embedding. Sentence transformers belongs to BERT family. It's an encoder only model. Optimized for semantic similarity and retrieval. Can be a better candidate for local RAG app, not enterprise RAG.

```python
train_embeddings = model.encode(data["train"]["text"], show_progress_bar=True)
```

```python
train_embeddings.shape
```

```python
train_embeddings[0]
```

```python
test_embeddings = model.encode(data["test"]["text"], show_progress_bar=True)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
clf = LogisticRegression(random_state=42)
```

```python
clf.fit(train_embeddings, data["train"]["label"]) # X_train, y_train
```

```python
y_pred = clf.predict(test_embeddings) # X_test
```

```python
evaluate_performance(data["test"]["label"], y_pred) # y_true, y_pred
```

##### Using Text-to-text transfer transformer models (T5)

<img src="./images/genai_65.png" width=1000 />

These models were first pretrained using masked language modelling. Instead of masking individual tokens, sets of tokens (or token spans) were masked during pretraining.

<img src="./images/genai_66.png" width=1000 />

Second steps is fine-tuning the base model. Instead of fine-tuning the model for one specific task, each task is converted to a sequence-to-sequence tasks and trained simultaneously.

<img src="./images/genai_67.png" width=1000 />

```python
pipe = pipeline(
    "text2text-generation",
    model="google/flan-t5-small",
    device='mps'
)
```

##### How about decoder only model ? like GPT-2 (which is open-source)

##### All GPT models with versions >=3 are closed-source. One has to create a billing account to retrieve the API keys.

```python
import torch
from transformers import GPT2Tokenizer, GPT2LMHeadModel

device = "mps" if torch.backends.mps.is_available() else "cpu"
print("Using device:", device)

model_name = "gpt2"
tokenizer = GPT2Tokenizer.from_pretrained(model_name)
tokenizer.pad_token = tokenizer.eos_token  # GPT-2 has no pad, reuse EOS

model = GPT2LMHeadModel.from_pretrained(model_name).to(device)
model.eval()

# -------- MOVIE REVIEW INPUT --------
# review = "What a movie ? Can you watch it without getting distracted ? Definitely yes but with eyes closed"
review = "Pls dont watch this movie. It is an utter waste of time"

# Few-shot prompt to teach GPT-2 what we want
prompt = (
    "Review: This movie was awful and boring.\n"
    "Sentiment: negative\n\n"
    "Review: I absolutely loved this film, it was amazing.\n"
    "Sentiment: positive\n\n"
    f"Review: {review}\n"
    "Sentiment:"
)

# Tokenize
inputs = tokenizer(prompt, return_tensors="pt").to(device)
print(inputs["input_ids"].shape)
input_len = inputs["input_ids"].shape[1]

with torch.no_grad():
    output_ids = model.generate(
        **inputs,
        max_new_tokens=3,      # we only expect 'positive' or 'negative'
        do_sample=False,       # greedy = deterministic
        pad_token_id=tokenizer.eos_token_id,
        eos_token_id=tokenizer.eos_token_id,
    )

# Only decode the NEW tokens after the prompt
generated_ids = output_ids[0, input_len:]
generated_text = tokenizer.decode(generated_ids, skip_special_tokens=True).strip()

# Clean up prediction to just 'positive' or 'negative'
prediction = generated_text.split()[0].lower()
if prediction.endswith("."):
    prediction = prediction[:-1]

print("\n=== MODEL RAW CONTINUATION ===")
print(repr(generated_text))

print("\n=== SENTIMENT PREDICTION ===")
print(prediction)
```

##### How is T5 compared to GPT-2 ?

```python
import torch
from transformers import GPT2Tokenizer, GPT2LMHeadModel

device = "mps" if torch.backends.mps.is_available() else "cpu"
print("Using device:", device)

model_name = "gpt2"

tokenizer = GPT2Tokenizer.from_pretrained(model_name)
model = GPT2LMHeadModel.from_pretrained(model_name).to(device)

input_text = "Explain machine learning"
inputs = tokenizer(input_text, return_tensors="pt").to(device)

output = model.generate(
    **inputs,
    max_length=150,
    do_sample=True,
    top_p=0.9,
    temperature=0.7,
)
print(tokenizer.decode(output[0], skip_special_tokens=True))
```

```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

device = "mps" if torch.backends.mps.is_available() else "cpu"
print("Using device:", device)

# You can use: "t5-small", "t5-base", etc.
# But FLAN-T5 is better for instruction-style prompts:
model_name = "google/flan-t5-base"

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name).to(device)

question = "Explain machine learning."

# T5 is encoder-decoder: we encode the input text, then generate
inputs = tokenizer(question, return_tensors="pt").to(device)

output_ids = model.generate(
    **inputs,
    max_new_tokens=2000,
    do_sample=True,
    top_p=0.9,
    temperature=0.7,
)

answer = tokenizer.decode(output_ids[0], skip_special_tokens=True)
print(answer)
```
